### MODIFY Phase

In the Modify phase, we clean the text data, encode sentiment labels into numerical format, and perform feature engineering with TF-IDF vectorization and additional features like text length.

```python
import re
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import TfidfVectorizer

# Drop missing values in 'text' column
df_semma = df_semma.dropna(subset=['text'])

# Define text cleaning function
def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\S+|https\S+", '', text, flags=re.MULTILINE)
    text = re.sub(r'\@\w+|\#', '', text)  # Remove mentions and hashtags
    text = re.sub(r"[^\w\s]", '', text)   # Remove punctuation
    text = text.strip()
    return text

df_semma['cleaned_text'] = df_semma['text'].apply(clean_text)

# Encode sentiment labels
label_encoder = LabelEncoder()
df_semma['sentiment_encoded'] = label_encoder.fit_transform(df_semma['sentiment'])

# TF-IDF Vectorization
tfidf_vectorizer = TfidfVectorizer(max_features=1000)
tfidf_features = tfidf_vectorizer.fit_transform(df_semma['cleaned_text'])

# Add text length feature
df_semma['text_length'] = df_semma['cleaned_text'].apply(len)

# Display processed data
df_semma[['cleaned_text', 'sentiment', 'sentiment_encoded', 'text_length']].head()
```
