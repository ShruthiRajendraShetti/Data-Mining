### Pre-Processing Phase

In the preprocessing phase, we clean the text data to remove noise, such as URLs and special characters, and we also encode sentiment labels into numerical format for further processing.

```python
import re
from sklearn.preprocessing import LabelEncoder

# Define text cleaning function
def kdd_clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\S+|https\S+", '', text, flags=re.MULTILINE)
    text = re.sub(r'\@\w+|\#', '', text)
    text = re.sub(r"[^\w\s]", '', text)
    text = text.strip()
    return text

# Apply cleaning
df_kdd['cleaned_text'] = df_kdd['text'].apply(kdd_clean_text)

# Encode sentiment labels
label_encoder = LabelEncoder()
df_kdd['sentiment_encoded'] = label_encoder.fit_transform(df_kdd['sentiment'])
df_kdd[['text', 'cleaned_text', 'sentiment', 'sentiment_encoded']].head()
```
