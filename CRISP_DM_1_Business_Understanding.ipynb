**Project Objective**: To classify tweets by sentiment - positive, negative, or neutral.
    
**Business Goals**:
1. Sentiment Monitoring
2. Customer Satisfaction
3. Market Research and Trend Analysis

**Success Criteria**:
- High accuracy in sentiment prediction (target: 80%+).
- Consistency across all sentiment types.
- Actionable insights for stakeholders.

**Key Assumptions and Constraints**:
1. Assumptions:
   - Tweets reflect genuine user sentiment.
2. Constraints:
   - Class imbalance and language complexity in tweets.
3. Risks:
   - Data quality and bias could impact prediction accuracy.
