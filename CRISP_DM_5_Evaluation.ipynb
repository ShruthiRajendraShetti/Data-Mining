### Evaluation Phase

**Evaluation Summary**:
- Accuracy: ~68%
- Best F1-score achieved for positive sentiment, lowest for negative.
- Limitations: Class imbalance and lack of nuanced capture by logistic regression.

**Improvements**:
- Experiment with advanced models (SVM, Random Forest).
- Address class imbalance through weighted classes or oversampling.
- Experiment with n-grams or word embeddings for richer text features.
